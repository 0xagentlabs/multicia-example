# Agent Token Vault orders
This notebook creates and validates order policy JSON without embedding private keys. A production agent injects a hardware/KMS-backed Solana `Signer` into `executePolicy`.


```python
import json
from pathlib import Path
policy = json.loads(Path('../automation/policy.example.json').read_text())
policy['orders'][0]['amount'] = '500000'
assert int(policy['orders'][0]['amount']) <= int(policy['orders'][0]['maxAmount'])
Path('../automation/policy.json').write_text(json.dumps(policy, indent=2))
policy
```
